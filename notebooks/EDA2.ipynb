```python
# demand_count=md['City'].value_counts()
# demand_company=md['Company'].value_counts()
demand=md.groupby(['Company','City']).size().reset_index(name='Total_Orders')
stores=md.groupby(['Company','City'])['No_of_Dark_Stores'].first().reset_index()
comparison=demand.merge(stores,on=['Company','City'])
valid_comparison = comparison[
    comparison['No_of_Dark_Stores'] > 0
]
valid_comparison['Demand_per_Store'] = (
    valid_comparison['Total_Orders'] /
    valid_comparison['No_of_Dark_Stores']
)
valid_comparison.sort_values(
    by='Demand_per_Store',
    ascending=False
).head(10)
```

```python
top10 = valid_comparison.sort_values(
    by='Demand_per_Store',
    ascending=False
).head(10)
plt.figure(figsize=(10,6))

plt.barh(
    top10['Company'] + " - " + top10['City'],
    top10['Demand_per_Store']
)

plt.xlabel("Demand per Store",fontsize=12)
plt.ylabel("Company & City",fontsize=12)
plt.title("Top Overloaded Dark Store Regions",fontsize=14,fontweight='bold')

plt.gca().invert_yaxis()
plt.grid(axis='x',alpha=0.3)
plt.show()
```

```python
analysis=md.groupby(
        ['Company','City']).agg({
        'Order_ID':'count',
        'No_of_Dark_Stores':'first',
        'Delivery_Time_Min':'mean'
}).reset_index()
```

```python
analysis.columns=[
    'Company',
    'City',
    'Total_Orders',
    'Stores',
    'Avg_Delivery_Time'
]
```

```python
analysis=analysis[analysis['Stores']>0]
analysis['Demand_Per_Store']=(
    analysis['Total_Orders']/analysis['Stores']
)
```

```python


# Create bubble chart
plt.figure(figsize=(12,8))

for company in analysis['Company'].unique():
    
    subset = analysis[analysis['Company'] == company]
    
    plt.scatter(
        subset['Demand_Per_Store'],
        subset['Avg_Delivery_Time'],
        
        # Bubble size based on total orders
        s=subset['Total_Orders'] / 2,
        
        alpha=0.8,
        label=company
    )

# Add labels and title
plt.xlabel("Demand Per Store", fontsize=12)
plt.ylabel("Average Delivery Time (min)", fontsize=12)

plt.title(
    "Operational Load vs Delivery Efficiency",
    fontsize=15,
    fontweight='bold'
)

# Legend
plt.legend(title="Company")

# Grid
plt.grid(alpha=0.3)

plt.show()
```

```python
Dark_Store_Impact=md.groupby('No_of_Dark_Stores')['Delivery_Time_Min'].mean()
Dark_Store_Impact
```

```python
md['store_group']=pd.cut(
    md['No_of_Dark_Stores'],
    bins=[0,1,3,5,100],
    labels=['No Dark Stores','1-2 Stores','3-5 Stores','6+ Stores']
)

dark_store_data=md.groupby('store_group')['Delivery_Time_Min'].mean()

labels=['No Dark Stores','1-2 Stores','3-5 Stores','6+ Stores']

values = [
    18.1,  # manually added for visualization
    dark_store_data.iloc[0],
    dark_store_data.iloc[1],
    dark_store_data.iloc[2]
]

plt.figure(figsize=(10,6))

plt.plot(
    labels,
    values,
    marker='o',
    linewidth=3
)

plt.title("Dark Store Impact on Delivery",fontweight='bold')
plt.xlabel("Number Of Dark Stores")
plt.ylabel("Delivery Time(Avg)")

plt.grid(alpha=0.3)
plt.show()
```

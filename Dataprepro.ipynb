```python
import pandas as pd
```

```python
df=pd.read_csv("cleaned_qcom_time.csv")
```

```python
df['Order_Time'].head(10)
```

```python
df['Order_Time']=pd.to_datetime(df['Order_Time'],errors='coerce')
```

```python
df['Order_Hour']=df['Order_Time'].dt.hour
```

```python
df['Order_Hour'].head(10)
```

```python
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
```

```python
df['Final_DateTime']=pd.to_datetime(df['Order_Date'].astype(str)+' ' +df['Order_Time'].dt.time.astype(str))
```

```python
df[['Order_Date', 'Order_Time', 'Final_DateTime']].head()
```

```python
df['Order_Month']=df['Final_DateTime'].dt.month_name()
```

```python
df['Order_Month']
```

```python
df['Order_Day']=df['Final_DateTime'].dt.day_name()
```

```python
df['Order_Day']
```

```python
df['Peak_Time']=df['Order_Hour'].apply(lambda x:'Peak' if 18 <= x >= 22 else 'Non Peak')
```

```python
df['Peak_Time']
```

```python
df['Peak_Time'].value_counts()
```

```python
df.drop('Order_Time',axis=1,inplace=True)
```

```python
df.columns
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
```

```python
df.columns
```

```python
df.to_csv('updated_qcom.csv',index=False)
```
